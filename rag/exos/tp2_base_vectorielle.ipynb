### Tâche 2.3 : Créer une collection

### Tâche 2.4 : Ajouter les documents à la collection

Ajoutez tous les documents avec leurs métadonnées à la collection.

```python
# TODO : Préparer et ajouter les documents
# 1. Extraire les textes, IDs et métadonnées de documents_data
# 2. Utiliser collection.add() avec documents, ids, metadatas
# Indices :
# - IDs : utiliser f"doc_{i}" pour chaque document
# - Documents : liste des textes
# - Metadatas : liste des dictionnaires de métadonnées

# Votre code ici
documents = [doc["text"] for doc in documents_data]
ids = [f"doc_{i}" for i in range(len(documents_data))]
metadatas = [
    {
        "category": doc["category"],
        "date": int(datetime.strptime(doc["date"], "%Y-%m-%d").timestamp()),
        "author": doc["author"],
        "level": doc["level"],
        "language": doc["language"],
    }
    for doc in documents_data
]

# Ajouter à la collection
collection.add(documents=documents, ids=ids, metadatas=metadatas)
```

## Partie 3 : Recherches Simples 

### Tâche 3.1 : Recherche basique

Effectuez une recherche simple sur "Python programmation" et affichez les 3 meilleurs résultats.

```python
# TODO : Effectuer une recherche simple
# Utiliser collection.query() avec :
# - query_texts : la requête
# - n_results : nombre de résultats

# Votre code ici
results = collection.query(query_texts=["Python programmation"], n_results=3)

print("Résultats bruts :")
print(f"Nombre de résultats : {len(results['documents'][0])}")
print(f"\nClés disponibles : {results.keys()}")
```

### Tâche 3.2 : Analyser les résultats

Créez une fonction pour afficher proprement les résultats de recherche.

```python
# TODO : Créer une fonction display_results(results)
# qui affiche de manière formatée :
# - Le rang du résultat
# - La distance/score
# - Le texte du document
# - Les métadonnées importantes (catégorie, niveau, auteur)

# Votre code ici
def display_results(results, title="Résultats de recherche"):
    print(f" {title}")

    if not results["documents"][0]:
        print("Aucun résultat trouvé")
        return

    for i in range(len(results["documents"][0])):
        doc = results["documents"][0][i]
        metadata = results["metadatas"][0][i]
        distance = results["distances"][0][i]
        doc_id = results["ids"][0][i]

        # Convertir distance en score de similarité (plus le score est élevé, mieux c'est)
        similarity = 1 - distance

        print(
            f"\n[{i + 1}] ID: {doc_id} | Similarité: {similarity:.4f} (distance: {distance:.4f})"
        )
        print("-" * 100)
        print(
            f"Catégorie: {metadata['category']} | Niveau: {metadata['level']} | Langage: {metadata['language']}"
        )
        print(f"Auteur: {metadata['author']} | Date: {metadata['date']}")
        print(f"\nTexte: {doc}")


# Afficher les résultats
display_results(results, "Recherche : Python programmation")
```

## Partie 4 : Recherches avec Filtres

### Tâche 4.1 : Filtre par catégorie

Recherchez "apprentissage" uniquement dans la catégorie "machine_learning".

```python
# TODO : Recherche avec filtre sur category
# Utiliser le paramètre where avec la syntaxe ChromaDB
# where={"category": "machine_learning"}

# Votre code ici
results_ml = collection.query(
    query_texts=["apprentissage"], n_results=5, where={"category": "machine_learning"}
)

display_results(results_ml, "Recherche : apprentissage [Filtre: machine_learning]")
```

### Tâche 4.2 : Filtres combinés

Recherchez "Python" avec les conditions :
- Niveau : intermédiaire OU avancé
- Date : après le 1er février 2024

```python
# TODO : Recherche avec filtres multiples
# Utiliser les opérateurs $or, $and, $gte
# Syntaxe ChromaDB :
# where={
#     "$and": [
#         {"$or": [{"level": "intermediate"}, {"level": "advanced"}]},
#         {"date": {"$gte": "2024-02-01"}}
#     ]
# }

# Votre code ici
results_advanced = collection.query(
    query_texts=["Python"],
    n_results=5,
    where={
        "$and": [
            {"$or": [{"level": "intermediate"}, {"level": "advanced"}]},
            {
                "date": {
                    "$gte": int(datetime.strptime("2024-02-01", "%Y-%m-%d").timestamp())
                }
            },
        ]
    },
)

display_results(
    results_advanced,
    "Recherche : Python [Filtres: (intermediate OR advanced) AND date >= 2024-02-01]",
)
```

### Tâche 4.3 : Filtre par auteur

Listez tous les documents de "Dr. Ahmed Khalil" sur le machine learning.

```python
# TODO : Recherche avec filtre auteur + catégorie
# Combiner where={"author": "...", "category": "..."}

# Votre code ici
results_author = collection.query(
    query_texts=["machine learning"],
    n_results=10,
    where={"$and": [{"author": "Dr. Ahmed Khalil"}, {"category": "machine_learning"}]},
)

display_results(
    results_author,
    "Recherche : machine learning [Filtres: auteur='Dr. Ahmed Khalil' AND category='machine_learning']",
)
```

### Tâche 5.1 : Expérience comparative

Pour la requête "conteneurs et déploiement", comparez :
1. Les résultats sans filtre
2. Les résultats avec filtre category="devops"
3. Les résultats avec filtre category="devops" ET level="beginner"

```python
# TODO : Comparer les trois recherches
# 1. Sans filtre
# 2. Avec filtre catégorie
# 3. Avec filtres catégorie + niveau

# Votre code ici
query = "conteneurs et déploiement"

# 1. Sans filtre
results_no_filter = collection.query(query_texts=[query], n_results=5)

# 2. Avec filtre catégorie
results_category_filter = collection.query(
    query_texts=[query], n_results=5, where={"category": "devops"}
)

# 3. Avec filtres catégorie + niveau
results_double_filter = collection.query(
    query_texts=[query],
    n_results=5,
    where={"$and": [{"category": "devops"}, {"level": "beginner"}]},
)

display_results(results_no_filter, "1. SANS FILTRE")
display_results(results_category_filter, "2. AVEC FILTRE CATÉGORIE (devops)")
display_results(results_double_filter, "3. AVEC DOUBLE FILTRE (devops + beginner)")
```
