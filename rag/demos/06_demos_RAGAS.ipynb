### Métriques RAGAS

1. **faithfulness** : La réponse est-elle fidèle au contexte fourni ?
2. **Answer revelancy** : La réponse répond-elle à la question ?
3. **Context Precision** : Les documents retournés sont-ils pertinents ?
4. **Context Recall** : Les informations neccessaires sont-elles présentes ?

```python
mistral_api_key = os.environ.get("MISTRAL_API_KEY")

llm = ChatMistralAI(model="mistral-small-latest", api_key=mistral_api_key)

embeddings = HuggingFaceEmbeddings(model_name="sentence-transformers/all-MiniLM-L6-v2")

run_config = RunConfig(timeout=120, max_workers=1)

metrics = [faithfulness, context_precision, context_recall, answer_relevancy]

results = evaluate(
    llm=llm,
    dataset=dataset,
    metrics=metrics,
    embeddings=embeddings,
    run_config=run_config,
)
```
